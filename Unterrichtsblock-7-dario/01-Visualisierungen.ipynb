```python
import pandas as pd
import plotly.graph_objects as go

df = pd.read_csv("../data/data_resampled.csv", index_col=0, parse_dates=True)
df
```

### Aufgabe 1
Filtere den Datensatz auf Raum 415 und visualisiere die Messreihe `co2` mit Plotly. Beschreibe danach die wichtigsten Muster oder Auffälligkeiten.


```python
df_415 = df[df["room_id"] == 415].copy()
df_415 = df_415.sort_index()
df_415["co2"] = df_415["co2"].mask(df_415["co2"] == -1)

fig = go.Figure()
fig.add_trace(go.Scatter(
    x=df_415.index,
    y=df_415["co2"],
    mode="lines",
    name="CO₂ Raum 415"
))
fig.update_layout(
    title="CO₂-Konzentration Raum 415",
    xaxis_title="Zeit",
    yaxis_title="CO₂ [ppm]"
)
fig.show()
```
