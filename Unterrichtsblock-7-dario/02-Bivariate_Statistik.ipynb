### Aufgabe 1
Der durchschnittliche CO2-Messwert pro Raum ist in `avg_co2_by_room` bereits vorbereitet. Erstelle daraus einen Barplot mit beiden Räumen.

```python
df = pd.read_csv("../data/data_resampled.csv", parse_dates=True)
df["room_id"] = df["room_id"].astype("string")  # damit die x-Achse nicht als Zahl interpretiert wird

# Durchschnittlicher CO2-Wert pro Raum
avg_co2_by_room = df.groupby("room_id")["co2"].mean()
avg_co2_by_room
```

```python
# Dein Code hier
fig = go.Figure()
fig.add_trace(go.Bar(
    x=avg_co2_by_room.index,
    y=avg_co2_by_room.values,
))
fig.update_layout(
    title="Durchschnittliche CO₂-Konzentration pro Raum",
    xaxis_title="Raum",
    yaxis_title="CO₂ [ppm]"
)
fig.show()
```

### Aufgabe 2
Visualisiere `temperature` und `humidity` aus `../data/room_413.csv` als Punktewolke. Prüfe, ob wärmere Phasen tendenziell trockener sind, und leite eine betriebliche Folgerung für den Raum ab.


```python
# Dein Code hier
df_413 = pd.read_csv("../data/room_413.csv", index_col=0, parse_dates=True)

# Fehlende Messwerte (-1) entfernen
df_413 = df_413[(df_413["temperature"] != -1) & (df_413["humidity"] != -1)]

fig = go.Figure()
fig.add_trace(go.Scatter(
    x=df_413["temperature"],
    y=df_413["humidity"],
    mode="markers",
    marker=dict(size=5, opacity=0.6)
))
fig.update_layout(
    title="Temperatur vs. Luftfeuchtigkeit Raum 413",
    xaxis_title="Temperatur [°C]",
    yaxis_title="Luftfeuchtigkeit [%]"
)
fig.show()
```
