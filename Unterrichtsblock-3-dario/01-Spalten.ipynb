# Aufgabe 7
Arbeite mit dem Datensatz `../data/data_resampled.csv`.

Mit `shift(1)` wird eine Spalte um eine Zeile nach unten verschoben. So kannst du den aktuellen `co2`-Wert direkt mit dem vorherigen Messwert vergleichen.

- a) Filtere den Datensatz auf den Raum `413`.
- b) Erstelle mit `shift(1)` eine neue Spalte `co2_aenderung`, welche die Veränderung zum vorherigen `co2`-Messwert enthält.
- c) Interpretiere das Vorzeichen der neuen Spalte: Was bedeuten positive und negative Werte?

Hinweis:
- Doku: [pandas.Series.shift](https://pandas.pydata.org/docs/reference/api/pandas.Series.shift.html)
- Der erste Wert darf leer sein, weil es keinen vorherigen Messwert gibt.

Wenn du willst, kannst du das Resultat auch kurz mit folgendem Code visualisieren. Plotting und Visualisierung schauen wir uns näher im Unterrichtsblock 7 an. 
```python
raum_413[["timestamp", "co2_aenderung"]].plot(
    x="timestamp",
    y="co2_aenderung",
    kind="line",

```

````python
df = pd.read_csv("../data/data_resampled.csv")
````

````python
df=df[df["room_id"] == 413]
````

````python
df["co2_aenderung"]=df["co2"]-df["co2"].shift(1)
````

````python
df[["timestamp", "co2", "co2_aenderung"]].head()
````
